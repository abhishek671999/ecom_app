```python
# You are provided with a CSV file containing 9 columns without headers.
# Your task is to:
# Extract only the 2nd and 5th columns from the file.
# Apply the following validations:
 
# 2nd Column: Must be numeric and contain exactly 4 digits.
# 5th Column: Must be alphanumeric and have a length of exactly 5 characters.
 
# Output only the rows that satisfy both validation conditions.

row_count = 0
delimiter = ','
output_data_list = []

def validate_numeric_values(text, length=4):
    if type(text) == 'int' and len(str(text)) == length:
        return True
    else:
        return False

def validate_alphanumeric_values(text, length=5):
    if type(text) == 'str' and len(str(text)) == length:
        return True
    else:
        return False


with open('test.csv', 'r') as fp:
    for line in fp.readline():
        line_list = line.split(delimiter)
        if row_count == 0:
            header = [line_list[1], line_list[4]]
        else:
            if validate_numeric_values(line_list[1]) & validate_alphanumeric_values(line_list[4]):
                output_data_list.append([line_list[1], line_list[5]])

df = pd.DataFrame(output_data_list, header=header)
df.to_csv('output.csv')
```
