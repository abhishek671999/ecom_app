```python
names = ["Alice", "Bob", "Charlie", "Diana", "Eve", "Frank", "Grace", "Henry", "Iris", "Jack"]
locations = ["New York", "Los Angeles", "Chicago"]
```

```python

num_of_rows = 1_000_000_000
names_array = F.array([F.lit(n) for n in names])
locations_array = F.array([F.lit(l) for l in locations])

df = spark.range(0, num_of_rows) \
    .withColumnRenamed("id", "user_id") \
    .withColumn("name", F.element_at(names_array, (F.floor(F.rand() * len(names)) + 1).cast("int"))) \
    .withColumn("location", F.element_at(locations_array, (F.floor(F.rand() * len(locations)) + 1).cast("int")))
```

```python
df.write.mode('overwrite').format('parquet').save('data/user_name_baseline.parquet')
```
